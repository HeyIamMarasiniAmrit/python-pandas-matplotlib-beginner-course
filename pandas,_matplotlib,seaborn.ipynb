```
#Bar Graph
import matplotlib.pyplot as plt

# Define data points
x = ['A', 'B', 'C', 'D']  # Categories
y = [5, 6, 7, 8]          # Values


plt.bar(x, y, color='skyblue', edgecolor='black')


plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Vertical Bar Graph")


plt.show()
```

```
import matplotlib.pyplot as plt
import numpy as np


labels = np.array(['A', 'B', 'C', 'D'])
sizes = np.array([53, 8, 1, 10])


colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']


plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=colors, startangle=140)


plt.title("Pie Chart Distribution")

plt.show()
```

```
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


np.random.seed(42)
df1 = pd.DataFrame({
    'bill_length_mm': np.random.uniform(35, 55, 50),
    'flipper_length_mm': np.random.uniform(170, 230, 50)
})


sns.lineplot(x='bill_length_mm', y='flipper_length_mm', data=df1)


plt.show()
```

```
import matplotlib.pyplot as plt
import seaborn as sns


df1 = sns.load_dataset("penguins")
df1 = df1.dropna(subset=['sex'])
sns.set_theme(style="whitegrid")
sns.barplot(x='island', y='bill_length_mm', data=df1, hue='sex', palette="muted")


plt.show()
```
