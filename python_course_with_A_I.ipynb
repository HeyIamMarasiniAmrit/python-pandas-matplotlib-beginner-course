```
# slicing
text = "Python"

# Get characters from index 0 up to (but excluding) 4
print(text[0:4])   # Output: Pyth

# If you omit the start, it starts from the beginning (0)
print(text[:3])    # Output: Pyt

# If you omit the stop, it goes all the way to the end
print(text[2:])    # Output: thon
```
